Sistem ini tidak hanya membalas input, tetapi juga menyajikan persona asisten virtual yang relevan dengan budaya lokal, lengkap dengan penanganan error (jaring pengaman fallback) yang responsif dan berkarakter.

Sorotan Teknis:

Teknik Refleksi ELIZA: Menerapkan ekstraksi teks (capture groups) untuk memantulkan kembali pernyataan pengguna, menciptakan ilusi empati dan pemahaman mesin.

Sistem Logika Hierarkis: Memanfaatkan urutan dictionary yang terstruktur untuk mencegah tabrakan pola (pattern collision) pada input pengguna yang kompleks.

```python
import re
import random

def ngobrol_sama_putri(input_user):
    # Mengubah input jadi huruf kecil semua biar pencocokan regex lebih gampang
    teks = input_user.lower()

    # Proses pencocokan (Pattern Matching) mencari pola di dalam dictionary
    for pola, daftar_jawaban in aturan_putri_full.items():
        cocok = re.search(pola, teks)

        if cocok:
            # Ambil hasil tangkapan (capture group) dari (.*) jika ada
            tangkapan = cocok.groups()

            # Pilih salah satu jawaban secara acak agar jawaban tidak kaku
            jawaban_terpilih = random.choice(daftar_jawaban)

            # Jika ada teks yang ditangkap (refleksi), masukkan ke dalam jawaban
            if tangkapan:
                return jawaban_terpilih.format(*tangkapan)
            return jawaban_terpilih

    # JARING PENGAMAN (Fallback) kalau user ngetik kata di luar kamus aturan
    fallback = [
        'Hah? Piye toh maksudmu? PUTRI ndak paham 😭',
        'Coba ngomongnya yang jelas dong, bingung nih PUTRI 😐',
        'Ooh gitu... terus terus? Kepo nih 🥹',
        'Walah, mboh ah. PUTRI lagi loading ini 🫠'
    ]
    return random.choice(fallback)
```
